```python
import akshare as ak
import pandas as pd
from pathlib import Path

def fetch_daily_sina(symbol: str, start: str, end: str, adjust: str = "hfq"):
    # 新浪接口需要带市场前缀，如 sh600519 或 sz000001
    if symbol.startswith('6'):
        full_symbol = f"sh{symbol}"
    else:
        full_symbol = f"sz{symbol}"
    
    # 1. 调用更新后的 stock_zh_a_hist_sina 接口
    df = ak.stock_zh_a_hist_sina(symbol=full_symbol, start_date=start, end_date=end, adjust=adjust)
    
    # 2. 将返回的英文列名统一重命名，并处理日期格式
    # 新浪接口返回的列名是英文，如 date, open, close 等
    df = df.rename(columns={"date": "日期", "close": "收盘"})
    df["date"] = pd.to_datetime(df["date"])
    return df.sort_values("date").reset_index(drop=True)

if __name__ == "__main__":
    out = Path("data/raw")
    out.mkdir(parents=True, exist_ok=True)
    
    for s in ["600519", "000001", "601318"]:
        print(f"正在从新浪抓取 {s} 的数据...")
        try:
            df = fetch_daily_sina(s, "20200101", "20261231")
            df.to_csv(out / f"{s}.csv", index=False)
            print(f"{s} 数据已保存！")
        except Exception as e:
            print(f"抓取 {s} 失败: {e}")
```
