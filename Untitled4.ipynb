```python
import pandas as pd
from pathlib import Path

# 1. 写一个通用的数据质量检查函数
def quality_report(df):
    return {
        "shape (行, 列)": df.shape,
        "missing_rate (缺失率)": df.isna().mean().round(4).to_dict(),
        "dup_rows (重复行数)": int(df.duplicated().sum()),
        # 额外检查“股票代码+日期”是否有重复 (baostock里代码列叫 code)
        "dup_code_date": int(df.duplicated(subset=["code", "date"]).sum()),
        "date_min (最早日期)": str(df["date"].min()),
        "date_max (最晚日期)": str(df["date"].max()),
    }

# 2. 读取我们刚才抓的茅台数据
df_test = pd.read_csv("data/raw/600519.csv")

# 3. 打印报告
report = quality_report(df_test)
for key, value in report.items():
    print(f"{key}: {value}")

# 4. 把结果存入 docs 文件夹（按教程要求）
out_dir = Path("docs")
out_dir.mkdir(parents=True, exist_ok=True)
report_file = out_dir / "data_quality_report.md"

# 写入 Markdown 文件
with open(report_file, "w", encoding="utf-8") as f:
    f.write("# 数据质量报告草稿\n\n")
    f.write(f"检查文件：data/raw/600519.csv\n\n")
    for key, value in report.items():
        f.write(f"- **{key}**: {value}\n")

print(f"\n报告已成功写入 {report_file}")
```
